```python
import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('Karnataka', 2010),
    ('Karnataka', 2020),
    ('Maharashtra', 2010),
    ('Maharashtra', 2020),
    ('Tamil Nadu', 2010),
    ('Tamil Nadu', 2020)
], names=['State', 'Year'])

population = pd.Series(
    [611, 675, 1124, 1244, 721, 778],
    index=index
)

print(population)
```

```python
print("Population of Karnataka:")
print(population.loc['Karnataka'])

print("Population in 2020:")
print(population.xs(2020, level='Year'))

print("Population of Maharashtra in 2020:")
print(population.loc[('Maharashtra', 2020)])
```

```python
df = population.unstack()

print("DataFrame using unstack():")
print(df)

s = df.stack()

print("Series using stack():")
print(s)
```

```python
import pandas as pd
import numpy as np

index = pd.MultiIndex.from_product(
    [['CSE', 'AIML'], [2023, 2024]],
    names=['Department', 'Year']
)

df = pd.DataFrame(
    np.random.randint(1, 100, size=(4, 2)),
    index=index,
    columns=['Students', 'Faculty']
)

print(df)
```

```python
import pandas as pd
import numpy as np

row_index = pd.MultiIndex.from_product(
    [['Python', 'DBMS'], ['Ravi', 'Priya']],
    names=['Subject', 'Student']
)

column_index = pd.MultiIndex.from_product(
    [['Term1', 'Term2'], ['Internal', 'External']],
    names=['Term', 'Exam_Type']
)

df = pd.DataFrame(
    np.random.randint(40, 100, size=(4, 4)),
    index=row_index,
    columns=column_index
)

print(df)
```

```python
print("Marks for Python:")
print(df.loc['Python'])

print("Mean marks grouped by Term:")
print(df.T.groupby(level='Term').mean().T)
```

```python
import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('StateB', 2020),
    ('StateA', 2010),
    ('StateC', 2020),
    ('StateB', 2010),
    ('StateA', 2020)
], names=['State', 'Year'])

data = pd.Series([20, 10, 30, 15, 25], index=index)

print("Original Series:")
print(data)

print("After sorting:")
data = data.sort_index()
print(data)

print("Partial Slice:")
print(data.loc['StateA':'StateB'])
```

```python
import pandas as pd

df = pd.DataFrame({
    'state': ['Karnataka', 'Karnataka', 'Tamil Nadu', 'Tamil Nadu'],
    'year': [2010, 2020, 2010, 2020],
    'population': [611, 675, 721, 778]
})

print("Original DataFrame:")
print(df)

multi_df = df.set_index(['state', 'year'])

print("DataFrame with MultiIndex:")
print(multi_df)

flat_df = multi_df.reset_index()

print("DataFrame after reset_index():")
print(flat_df)
```
