```python
import numpy as np

x = np.arange(1, 13)
matrix = x.reshape(3, 4)

print("3x4 Matrix:")
print(matrix)
column1 = x.reshape((12, 1))
column2 = x[:, np.newaxis]

print("\nColumn vector using reshape:")
print(column1)

print("\nColumn vector using np.newaxis:")
print(column2)
print("\nAre both column vectors identical?", np.array_equal(column1, column2))
```

```python
import numpy as np

np.random.seed(0)

a = np.random.randint(0, 10, size=(2, 3))
b = np.random.randint(0, 10, size=(2, 3))

print("Array A:")
print(a)

print("\nArray B:")
print(b)

vstack_result = np.vstack((a, b))
concat_axis0 = np.concatenate((a, b), axis=0)

print("\nVertical using vstack:")
print(vstack_result)

print("\nVertical using concatenate:")
print(concat_axis0)

print("Vertical results are same:",
      np.array_equal(vstack_result, concat_axis0))

hstack_result = np.hstack((a, b))
concat_axis1 = np.concatenate((a, b), axis=1)

print("\nHorizontal using hstack:")
print(hstack_result)

print("\nHorizontal using concatenate:")
print(concat_axis1)

print("Horizontal results are same:",
      np.array_equal(hstack_result, concat_axis1))
```

```python
import numpy as np

arr = np.arange(1, 16)

print("Original array:")
print(arr)

parts = np.split(arr, 3)

print("\nPart 1:", parts[0])
print("Part 2:", parts[1])
print("Part 3:", parts[2])
```

```python
import numpy as np

arr = np.arange(1, 25).reshape(6, 4)

print("Original 6x4 array:")
print(arr)
horizontal = np.hsplit(arr, 2)

print("\nUsing hsplit():")
print("Part 1:")
print(horizontal[0])

print("Part 2:")
print(horizontal[1])
vertical = np.vsplit(arr, 2)

print("\nUsing vsplit():")
print("Part 1:")
print(vertical[0])

print("Part 2:")
print(vertical[1])
```
