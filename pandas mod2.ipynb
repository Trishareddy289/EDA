```python
import pandas as pd

population = pd.Series(
    [12400000, 16700000, 8400000, 7100000, 4500000],
    index=['Mumbai', 'Delhi', 'Bengaluru', 'Chennai', 'Kolkata']
)

print("Population Series:")
print(population)

print("Population of Bengaluru:")
print(population['Bengaluru'])

print("Cities from Delhi to Chennai:")
print(population['Delhi':'Chennai'])
```

```python
import pandas as pd

area = pd.Series(
    [307713, 191791, 155707, 130058, 112077],
    index=['Rajasthan', 'Karnataka', 'Andhra Pradesh',
           'Tamil Nadu', 'Kerala']
)

population = pd.Series(
    [800, 700, 530, 770, 350],
    index=['Rajasthan', 'Karnataka', 'Andhra Pradesh',
           'Tamil Nadu', 'Kerala']
)

states = pd.DataFrame({
    'area': area,
    'population': population
})

states['density'] = states['population'] / states['area']

print(states)
```

```python
import pandas as pd

indA = pd.Index([2, 3, 5, 7, 11])
indB = pd.Index([1, 3, 5, 7, 9])

print("Intersection:")
print(indA.intersection(indB))

print("Union:")
print(indA.union(indB))

print("Symmetric Difference:")
print(indA.symmetric_difference(indB))
```
