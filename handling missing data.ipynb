```python
import pandas as pd
import numpy as np

data = pd.Series([1, np.nan, 'hello', None])

print("Null Values:")
print(data.isnull())

print("Non-Null Values:")
print(data[data.notnull()])
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.arange(1, 21).reshape(5, 4),
                  columns=['A', 'B', 'C', 'D'])

df.loc[0, 'A'] = np.nan
df.loc[1, 'B'] = np.nan
df.loc[2, 'A'] = np.nan
df.loc[3, 'C'] = np.nan
df.loc[4, 'D'] = np.nan

print("DataFrame:")
print(df)

print("Missing Values:")
print(df.isnull().sum())
```

```python
print("After dropna():")
print(df.dropna())

print("After dropna(axis='columns'):")
print(df.dropna(axis='columns'))
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, np.nan, np.nan, 4],
    'B': [2, np.nan, 6, np.nan],
    'C': [np.nan, np.nan, 7, 8]
})

print("Original DataFrame:")
print(df)

print("After dropna(how='all'):")
print(df.dropna(how='all'))
```

```python
print("After dropna(thresh=3):")
print(df.dropna(thresh=3))
```

```python
import pandas as pd
import numpy as np

data = pd.Series([1, np.nan, 2, None, 3, np.nan])

print("Original Series:")
print(data)

print("Fill with 0:")
print(data.fillna(0))

print("Forward Fill:")
print(data.ffill())

print("Backward Fill:")
print(data.bfill())
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'Temperature': [30, 32, np.nan, 36, np.nan, 40, 42]
})

print("Original DataFrame:")
print(df)

df['Mean Filled'] = df['Temperature'].fillna(
    df['Temperature'].mean()
)

df['Interpolated'] = df['Temperature'].interpolate()

print("After Filling Missing Values:")
print(df)
```
