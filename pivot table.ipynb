```python
import pandas as pd

data = {
    'department': ['CSE', 'CSE', 'AIML', 'AIML', 'ECE', 'ECE',
                   'CSE', 'CSE', 'AIML', 'AIML', 'ECE', 'ECE'],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F',
               'F', 'M', 'F', 'M', 'F', 'M'],
    'salary': [50000, 60000, 55000, 65000, 45000, 50000,
               52000, 58000, 62000, 58000, 48000, 52000]
}

df = pd.DataFrame(data)

print("Employee Data:")
print(df)

pivot = pd.pivot_table(
    df,
    values='salary',
    index='gender',
    columns='department',
    aggfunc='mean'
)

print("\nAverage Salary Pivot Table:")
print(pivot)
```

```python
pivot_count = pd.pivot_table(
    df,
    values='salary',
    index='gender',
    columns='department',
    aggfunc='count'
)

print(pivot_count)
```

```python
pivot_total = pd.pivot_table(
    df,
    values='salary',
    index='gender',
    columns='department',
    aggfunc='mean',
    margins=True
)

print(pivot_total)
```

```python
df['age'] = [24, 32, 28, 45, 26, 38, 23, 35, 29, 42, 31, 48]

df['age_category'] = pd.cut(
    df['age'],
    bins=[0, 25, 40, 60],
    labels=['Junior', 'Mid', 'Senior']
)

pivot_age = pd.pivot_table(
    df,
    values='salary',
    index='age_category',
    columns='department',
    aggfunc='mean',
    observed=False
)

print(df[['age', 'age_category']])
print("\nAverage Salary by Age Category:")
print(pivot_age)
```

```python
pivot_stats = pd.pivot_table(
    df,
    index='department',
    values=['salary', 'age'],
    aggfunc={
        'salary': 'sum',
        'age': 'mean'
    }
)

print(pivot_stats)
```

```python
import seaborn as sns
import pandas as pd

titanic = sns.load_dataset('titanic')

pivot_survival = pd.pivot_table(
    titanic,
    values='survived',
    index='sex',
    columns='class',
    aggfunc='mean',
    margins=True,
    observed=False
)

print(pivot_survival)
```
