```python
import numpy as np
a = np.arange(1, 21)
print("Array:")
print(a)
print("Number of dimensions:", a.ndim)
print("Shape:", a.shape)
print("Size:", a.size)
print("Data type:", a.dtype)
```

```python
import numpy as np
arr = np.arange(1, 13).reshape(3, 4)
print("Array:")
print(arr)
print("Element in row 2, column 3:", arr[1, 2])
print("Last row:", arr[-1])
print("Last column:", arr[:, -1])
```

```python
import numpy as np
arr = np.arange(1, 13).reshape(3, 4)
print("Array:")
print(arr)
print("Subarray:")
print(arr[0:2, 0:2])
print("Alternate columns:")
print(arr[:, ::2])
print("Reverse columns:")
print(arr[:, ::-1])
```

```python
import numpy as np
arr = np.arange(1, 21).reshape(4, 5)
print("Original Array:")
print(arr)
view = arr[0:3, 0:4]
view[0, 0] = 100
print("\nOriginal Array after modifying view:")
print(arr)
copy = arr[0:3, 0:4].copy()
copy[0, 0] = 200
print("\nOriginal Array after modifying copy:")
print(arr)
```

```python
import numpy as np
x = np.arange(1, 13)
print("Original Array:")
print(x)
a = x.reshape(3, 4)
print("\n3x4 Matrix:")
print(a)
b = x.reshape(12, 1)
print("\nColumn Vector using reshape:")
print(b)
c = x[:, np.newaxis]
print("\nColumn Vector using newaxis:")
print(c)
print("\nBoth are identical:", np.array_equal(b, c))
```

```python
import numpy as np
np.random.seed(0)
a = np.random.randint(1, 10, (2, 3))
b = np.random.randint(1, 10, (2, 3))
print("Array A:")
print(a)
print("\nArray B:")
print(b)
v1 = np.vstack((a, b))
v2 = np.concatenate((a, b), axis=0)
print("\nVertical using vstack:")
print(v1)
print("\nVertical using concatenate:")
print(v2)
print("Vertical same:", np.array_equal(v1, v2))
h1 = np.hstack((a, b))
h2 = np.concatenate((a, b), axis=1)
print("\nHorizontal using hstack:")
print(h1)
print("\nHorizontal using concatenate:")
print(h2)
print("Horizontal same:", np.array_equal(h1, h2))
```

```python
import numpy as np
arr = np.arange(1, 16)
print("Original Array:")
print(arr)
parts = np.split(arr, 3)
print("\nPart 1:")
print(parts[0])
print("\nPart 2:")
print(parts[1])
print("\nPart 3:")
print(parts[2])
```

```python
import numpy as np
arr = np.arange(1, 25).reshape(6, 4)
print("Original Array:")
print(arr)
h = np.hsplit(arr, 2)
print("\nHorizontal Split:")
print(h[0])
print(h[1])
v = np.vsplit(arr, 2)
print("\nVertical Split:")
print(v[0])
print(v[1])
```
